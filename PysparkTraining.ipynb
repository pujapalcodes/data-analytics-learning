```python
from pyspark.sql.functions import unix_timestamp, round
```

```python
df1 = df.withColumn(
'trip_duration_minutes',
 round(
     unix_timestamp('tpep_dropoff_datetime')-(unix_timestamp('tpep_pickup_datetime'))/ 60,
     1))
```

```python
df1.printSchema()
```

```python
df2 = df1.select(
    'tpep_pickup_datetime',
    'tpep_dropoff_datetime',
    'trip_duration_minutes',
    'PULocationID',
    'DOLocationID',
    'passenger_count',
    'fare_amount',
    'Airport_fee',
    'total_amount'
) \
.withColumnRenamed("tpep_pickup_datetime", "pu_datetime") \
.withColumnRenamed("tpep_dropoff_datetime", "do_datetime") \
.withColumnRenamed("PULocationID", "pu_location_id") \
.withColumnRenamed("DOLocationID", "do_location_id") \
.withColumnRenamed("Airport_fee", "airport_fee")
```

```python
df2.show()
```

```python
#load the taxi ride data and create a temporary view
taxi_jan2026 = spark.read.parquet('/content/drive/MyDrive/Pyspark/yellow_tripdata_2026-01.parquet')
taxi_jan2026.createOrReplaceTempView('taxi_jan2026')
```

```python
#load the lookup data and create a temporary view
taxi_lookup = spark.read.option('header', True).csv('/content/drive/MyDrive/Pyspark/taxi_zone_lookup.csv')
```

```python
taxi_lookup.createOrReplaceGlobalTempView('taxi_lookup')
```

```python
#join the two tables using SQL and assign the result to a dataframe called joined_df
query ='''
SELECT
 DOLocationID,
 Borough,
 Total_amount
FROM taxi_jan2026
LEFT join global_temp.taxi_lookup
 on DOLocationID = LocationID
'''
joined_df = spark.sql(query)
```

```python
joined_df.groupBy('Borough').agg(avg('total_amount').alias('avg_amount')).show()
```
