```python
df1 = df.fillna({'passenger_count' : 1})
```

```python
df1.show()
```

```python
from pyspark.sql.functions import unix_timestamp, round
```

```python
df1 = df.withColumn(
'trip_duration_minutes',
 round(
     unix_timestamp('tpep_dropoff_datetime')-(unix_timestamp('tpep_pickup_datetime'))/ 60,
     1))
```

```python
df2 = df1.select('tpep_pickup_datetime', 'tpep_dropoff_datetime',\
               'trip_duration_minutes', 'PULocationID', 'DOLocationID',\
                 'passenger_count', 'fare_amount', 'Airport_fee', 'total_amount').\
withColumnsrenamed({'tpep_pickup_datetime': 'pu_datetime',
                   'tpep_dropoff_datetime' : 'do_datetime',\
                    'PULocationID' : 'pu_location_id',\
                    'DOLocationID': 'do_location_id',\
                    'Airport_fee' : 'airport_fee'})
```

```python
df1.printSchema()
```

```python
df2 = df1.select(
    'tpep_pickup_datetime',
    'tpep_dropoff_datetime',
    'trip_duration_minutes',
    'PULocationID',
    'DOLocationID',
    'passenger_count',
    'fare_amount',
    'Airport_fee',
    'total_amount'
) \
.withColumnRenamed("tpep_pickup_datetime", "pu_datetime") \
.withColumnRenamed("tpep_dropoff_datetime", "do_datetime") \
.withColumnRenamed("PULocationID", "pu_location_id") \
.withColumnRenamed("DOLocationID", "do_location_id") \
.withColumnRenamed("Airport_fee", "airport_fee")
```
